```python
# Trabalhando com funções

def verificar_maioridade(idade):
  if idade >= 18:
    return "Maior de Idade"
  else:
    return "Menor de Idade"

try:
  idade = int(input("Qual é a sua idade?\n"))
  print(verificar_maioridade(idade))

except ValueError:
  print("-" * 40)
  print("🚨 ERRO DE DIGITAÇÃO DETECTADO!")
  print("Por favor, use apenas números (Ex: 25).")
  print("-" * 40)
```

```python
from os import stat_result
# Trabalhando com funções

def calcular_parcela(valor_total, quantidade_parcelas):
  valor_parcelas = valor_total / quantidade_parcelas
  return valor_parcelas
def verificar_aprovacao(valor_da_parcela, salario_do_cliente):
  salario_porcentagem = salario_do_cliente * 0.3
  if salario_porcentagem > valor_da_parcela:
    return "Financiamento Aprovado!"
  else:
    return "Financiamento Negado!"
valor_carro = 60000
qnt_parcelas = 44
salario = 3000

resultado_parc = calcular_parcela(valor_carro, qnt_parcelas)
stat_result = verificar_aprovacao(resultado_parc, salario)
print(f"Valor da parcela: {resultado_parc:.2f}")
print(f"Status da Agencia: {stat_result}")
```

```python
 **🤖 TREINO DE AUTOMAÇÃO: Guardando os dados no arquivo**
```

```python
with open("relatorio_banco.txt", "a") as arquivo:

  arquivo.write(f"Valor das Parcelas R$ {resultado_parc}\nEstatus da Agencia: {stat_result}")

print("Automação concluída, dados injetados com sucesso!")
```
