```python
# Sistema Automatizado de Rh (Recrutamento)

import os
import shutil

def avaliar_candidato(nota):
    if nota >= (nota * 0.7):
      return "Aprovado para Entrevista"
    else:
      return "Reprovado para Entrevista"

# Passo 2: Use o try/except para capturar o input da nota com segurança
try:
    nome = input("Nome do candidato:\n")
    nota = int(input("Nota do teste técnico:\n"))
    print("-"*35)

    # Passo 3: Chame a função e guarde o resultado em uma variável
    resultado_candidato = avaliar_candidato(nota)
    print(f"Candidato: {nome}\nStatus: {resultado_candidato}")
    print("-"*35)

    # Passo 4: Grave o Nome e o Status no arquivo "triagem_candidatos.txt"
    with open("triagem_candidatos.txt", "a") as arquivo:
      arquivo.write(f"Candidato: {nome}\nStatus: {resultado_candidato}.")

except ValueError:
    print("Erro: Digite uma nota numérica válida!")

# Passo 5: Fora do try/except, use o os e shutil para criar a pasta "Processo_Seletivo" e mover o arquivo para lá
nome_pasta = "Processo_Seletivo"
if not os.path.exists(nome_pasta):
  os.mkdir(nome_pasta)
  print(f"Pasta {nome_pasta} criada com sucesso")
  print("-"*35)

shutil.move("triagem_candidatos.txt", "Processo_Seletivo/triagem_candidatos.txt")
print("Robô concluído com sucesso")
```
