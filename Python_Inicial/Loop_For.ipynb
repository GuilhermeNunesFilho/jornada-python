```python
# Loop for, usado para percorrer uma determinada lista
# sua estrutura base:
for i in range(5): # essa é a sua estrutura base "para cada i no intervalo de()
    print(i)
```

```python
num = 10
for i in range(num,0,-1): # podemos usar até 3 parametris dentro do for "inicio,fim e intervalo" sempre nessa ordem
    print(i)
```

```python
for i in range(5):
    print("Python") # Quando não tiver uma referencia, irá repetir oque estiver dentro do bloco for
```

```python
# Tabuada
num = int(input("didite um número"))
cont = 0
for i in range(num,num+11,+1):
    res = (num * cont)
    print(f"{num} x {cont} = {res}")
    cont += 1
```
