```python
def somar(): # def é a palavra chave para criar uma função 'def = definir'
    a = int(input('Digite um número: '))
    b = int(input('Digite outro número: '))
    res = (a+b)
    print(f"A soma de {a} com {b} é: {res}")
    return res # return é a palavra chave para retornar um valor da função, nesse caso irá retornar a variável res
```

```python
somar()
```

```python
def testando_maior_idade():
    ano_atual = 2026
    nome = input("Olá, qual é seu nome: ")
    print(f"Muito prazer!")
    ano = int(input(f"Que ano você nasceu {nome}"))
    idade = (ano_atual - ano)
    if (ano_atual - ano) >= 18:
        print(f"Parabéns {nome}, você já é maior!\nSua idade é: {idade}")
    else:
        print(f"Você tem apenas {idade} anos!\nAproveite sua menor idade..")
```

```python
testando_maior_idade()
```

```python
testando_maior_idade()
```

```python
def mult(x,y):
    res = x**y # '**' Operador matemático de elevação
    print(f"Resultado: {res}")
```

```python
print("----Multiplicação!----")
num1 = int(input("Digite um número: "))
num2 = int(input(f"Digite outro número p/ multiplicar com {x}: "))

mult(num1,num2) # essa função foi definida com 2 paramentros (x e y) por isso ela nessecita de 2 valores (num1 e num2)
```
