```python
# Função de pesquisa binária

def pesquisa_binaria(lista, item):
  baixo = 0
  alto = len(lista) -1

  while baixo <= alto:
    meio = (baixo + alto) // 2
    chute = lista[meio]

    if chute == item:
      return meio

    if chute > item:
      alto = meio - 1

    else:
      baixo = meio + 1
  return None
```

```python
minha_lista = [
    1,3,5,7,9,11,13,17
]
escolha = float(input("Escolha um primo:\nEx: 1,3,5,7,9,11,13,17\n"))

resultado = pesquisa_binaria(minha_lista,escolha)

print(f"Você escolheu {escolha}, sua resposta é: {resultado}")
```
