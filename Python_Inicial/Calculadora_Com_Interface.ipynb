```python
# 1. Importar as Bibliotecas

import tkinter as tk # Biblioteca para criar interfaces
```

```python
# 2. Criar funções 'limpar e calcular'

def click(num):
    entry.insert(tk.END, num) # Inserção de dados

def clear():
    entry.delete(0, tk.END) # Limpar a tela da calculadora 

def calculate():
    try: # Tratamento de erro caso os dados inseridos sejam incorretos. Ex: s + x...
        result = eval.entry(entry.get())
        entry.delete(0, tk.END)
        entry.insert(tk.END, str(result))
    except: # fechando tratamento, limpando o campo e exibindo msg de erro
        entry.delete(0, tk.END)
        entry.insert(tk.END, "Erro")
```

```python
# 3. Criar a janela de interface da calculadora

root = tk.Tk()
root.title("Calculadora Simples")
```

```python
# 4. Criar o campo de entradas e saída dos resutados

entry = tk.Entry(root, width = 20, borderwidth = 5, font = ("Arial", 14))
entry.grid(row = 0, column = 0, columnspan = 4)
```
