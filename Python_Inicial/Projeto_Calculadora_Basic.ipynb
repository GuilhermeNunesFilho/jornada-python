```python
# Criando uma função para cada operação
# Solicitando ao usuário que escolha uma operação
# Solicitando ao usuário que insira dois valores
# Realizando a operação escolhida pelo o usuário
```

```python
# Criando uma função para cada operação

def adicao(x,y):
    return x+y
def subtracao(x,y):
    return x-y
def mult(x,y):
    return x*y
def divisao(x,y):
    return x/y
```

```python
## Solicitando ao usuário que escolha uma operação
while True:

    print("-"*35)
    print("=======<CALCULADORA_BASICA>=======")
    print("Escolha uma Operação: \n 1: Adição\n 2: Subtração\n 3: Multiplicação\n 4: Divisão\n 5: Sair\n")
    opcao = int(input("Informe a Operação: "))
    print("-"*35)

    if opcao == 5:
        print("Encerrando...")
        break

    elif opcao in [1, 2, 3, 4]:
        num1 = int(input("Digite um número: "))
        num2 = int(input("Digite outro número: "))

        if opcao == 1:
            print("=======<SOMA>=======")
            print(f"Resultado: {adicao(num1,num2)}" )
        
        elif opcao == 2:
            print("=======<SUBTRAÇÃO>=======")
            print(f"Resultado: {subtracao(num1,num2)}" )
        
        elif opcao == 3:
            print("=======<MULTIPLICAÇÃO>=======")
            print(f"Resultado: {mult(num1,num2)}" )
        
        elif opcao == 4:
            print("=======<Divisão>=======")
            print(f"Resultado: {divisao(num1,num2)}" )
        
        else:
            print("Opção Invalida")
```
