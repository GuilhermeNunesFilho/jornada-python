```python
# Variáveis são espaço reservado da memória para armazenar um dado
```

```python
# Declarando os tipos de uma váriavel, sempre usando um nome sujestivo ao dado(Boa Prática)

# Variável tipo int
idade = 41 # Idade, é o nome que dei a variável e 41 seu dado armazenado através da atribuição '='

# Mostrando o valor da variável 'idade'
print(idade)
```

```python
# Variável tipo float
altura = 1.82 # Float é a variável com casas decimais(número quebrado), no entanto são separada por ponto ao invés de virgula
print(altura)
```

```python
# Variável tipo str ou string
aluno = "João" # Perceba que a variável tipo str, seu dado é posto entre áspa ""
print(aluno)
```

```python
# Variável tipo Bool(verdadeiro=True ou falso=False)
vdd = True
fals = False
print(vdd); print(fals) # Ao usar ponto e virgula em Python é o mesmo que pula uma linha
```

```python
# Brincando com as variáveis Bool
fals == False # Usamos os dois iguais '==' quando realmente queremos fazer a comparação igual
vdd == True
```
