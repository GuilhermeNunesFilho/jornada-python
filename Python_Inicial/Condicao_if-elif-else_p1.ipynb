```python
# A condição if, elif e else serve para uma tomada de decisão. Ex: "Se isso for verdade, aquilo será feito, coso ao contrário outra coisa será feita"
```

```python
# Vamos utilizar nesse cod o input
bebida = input("Bebida: ")
teor = int(input("Seu teor: "))

# Vamos usa a condição if

if teor == 0:
    print("{} é uma Bebida saúdavel".format(bebida))
```

```python
# Copiei o cod pra fica mais fácil o entendimento
bebida = input("Bebida: ")
teor = int(input("Seu teor: "))

# Vamos usa a condição if e elif
if teor == 0: # '==' é um operador lógico, são muitos usados em programação
    print("{} é uma Bebida saúdavel".format(bebida))

# Aqui vou usr um elif para um teor maior ou igual a 4
elif teor >= 4: # '>=' operador lógico de maior ou igual
    print(f"{bebida}, é uma Bebida leve, porém não é recomendado seu uso abusivel!") # Jeito simplificado de usar o .format
```

```python
# Vamos um ultimo elif
bebida = input("Bebida: ")
teor = int(input("Seu teor: "))

# Vamos usa a condição if, elif e else

if teor == 0:
    print("{} é uma Bebida saúdavel".format(bebida))

# Aqui vou usr um elif para um teor maior ou igual a 4
elif teor >= 4 and teor < 35: # '>=' operador lógico de maior ou igual, '<' operador menos, 'and' operador E
    print(f"{bebida}, é uma Bebida leve, porém não é recomendado seu uso abusivel!") # Jeito simplificado de usar o .format

# Vou repeti outro elif para um teor maior que 35
elif teor > 34: #'>' operador lógico maior
    print(f"{bebida}, autamente forte, não use!\nSeu uso pode acarretar em desordem social e até tragédias fatais.\nCuidado")
```
