```python
# Operadores MATEMÁTICOS EM PYTHON
# No print, quando trabalhamos com inteiros (números), não ha necessidade de usar áspas
```

```python
# Operador de Adição '+'
print(10 + 50)
```

```python
# Operador de Subtração '-'
print(80 - 30)
```

```python
# Operador de Divisão '/'
print(100 / 3)
```

```python
# Operador de Multipolicação '*'
print(10 * 100)
```

```python
# Operador de Potência '**'
print(10 ** 3)
```

```python
# Operador p/ extrair a parte inteira da divisão '//'
print(100 // 3)
```

```python
# Operador p/ extrair o resto da divisão '%'
print(10 % 3) # 10 dividido por 3 = a 3 e sobra '1'. Esse é o resto da divisão!
```
