```python
# While é um loop de um código que só será interrupido caso uma condição seja atendida
cont = 0
while cont < 3:
    print("Eu Amo Programar")
    cont += 1 # Esse incremento é fundamental para o fim do programa evitando seu loop infinito ('cont += 1' = 'cont = cont + 1' )
```

```python
# Criando uma calculadora de fatorial
numero = int(input("Digite um número: "))
fatorial = numero
cont = 1
while (numero - cont) > 1:
    fatorial = fatorial * (numero - cont)
    cont += 1
print(f"Número: {numero}\nFatorial: {fatorial}")
```
