```python
import pandas as pd
import numpy as np
```

```python
pares = pd.Series([2,4,6,8,42,60]) # Matriz unidimencional
print(pares)
```

```python
impares = np.array([1,3,5,7,9,43,101])  # Matriz unidimencional
print(impares)
```

```python
primos = np.array([(2,3,5,7),(11,13,17,19)]) # Duas Matriz é bidimensional, esse tipo de estrutura só aceita no Numpy
print(primos)
```

```python
# tentando usar o pandas para lê uma matriz bidimensional
sequencia = pd.Series(primos)
print(sequencia)
```

```python
sequencia1 = pd.Series(pares)
print(sequencia1)
```

```python
sequencia2 = pd.Series(impares)
print(sequencia2)
```
