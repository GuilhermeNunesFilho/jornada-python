```python
# Modo 'simples' para manipulação de lista
```

```python
dolar = [1,2,3,4,5,6,7]
# Mudando de lista e multiplicando os valores

real1=[]
for i in dolar:
    real1.append(i*5)
print(real1)
```

```python
# Usando a função map()
real2 = list(map(lambda x: x*5, dolar))
print(real2)
```

```python
real3 = list(map(lambda y: y*3, real1))
print(real3)
```
