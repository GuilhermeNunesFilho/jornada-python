```python
# Append é uma função usada para acrecentar itens a lista

lista_teste = [
    5,8,11,14
]

print(lista_teste)
```

```python
numeros_completar = [
    1,2,3,4,6,7,9,10,12,13
]

# Usarei um for para percorrer a lista numeros_completar
for i in numeros_completar:
    # função appende para adicionar a variável i, que está recebendo a lista numeros_completar, na lista_teste
    lista_teste.append(i)
print(lista_teste)
```

```python
# Ordenando os números da lista

ordenados = sorted(lista_teste)
print(ordenados)
```
