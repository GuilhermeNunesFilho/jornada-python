```python
def maiorMenor():
    x = int(input("Qual sua idade?"))
    y = 18
    if x >= y:
        print(f"Você tem {x} anos, já é maior de idade!")

    else:
        print(f"Você tem {x} anos, ainda é menor!")
    return
```
