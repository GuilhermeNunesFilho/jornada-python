```python
# Listas são como listas reais usadas no mundo fisico, ex: uma lista de produtos, nomes, tarefas etc...
```

```python
# Criando uma lista

lista1 = [1,2,3,4,5,6,7,8]

print(lista1)
type(lista1) # função para descobrimos o tipo de dado de um objeto
```

```python
# Listas Aninhada

lista2 = [
    [1,2,3,4], # indice 0
    [5,6,7,8], # indice 1
    [9,10,11,12] # indice 2
]

print(lista2)
type(lista2)
```

```python
# Busca em lista pelos os indice

print(lista2[1][2]) # primeiro coxetes é o indice da lista, o segundo do seu conteúdo
```
