```python
# Automação de Relatórios com Planilhas

import openpyxl

# 1. Cria um livro de planilhas em branco
wb = openpyxl.Workbook()

# 2. Seleciona a folha que já vem aberta (a folha ativa)
aba_ativa = wb["Sheet"]
aba_ativa.title = "Vendas"

# 3. Para escrever em uma célula específica, você usa a coordenada dela:
aba_ativa["A1"] = "Vendedor"
aba_ativa["B1"] = "Produto"
aba_ativa["C1"] = "Valor do Produto"
aba_ativa["D1"] = "Comissão (10%)"

aba_ativa["A2"] = "Guilherme"
aba_ativa["B2"] = "Curso de Python"
aba_ativa["C2"] = 100

# 4. Para salvar o arquivo no final:
aba_ativa["D2"] = aba_ativa["C2"].value * 0.1
wb.save("Automatizador_de_Comissões_de_Vendas.xlsx")
```

****Inserindo novas vendas em nossa tabela Excel Automatizada****

```python
novas_vendas = [
    ["Ana", "Computador", 5000],
    ["José", "Teclado Gamer", 200],
    ["Lucas", "Celular Google", 2000],
    ["João", "Celular IPhone", 10000]
]


for venda in novas_vendas:

  valor_produto = venda[2]

  comissao = valor_produto * 0.1

  linha_completa = [venda[0], venda[1], venda[2], comissao]

  aba_ativa.append(linha_completa)

aba_ativa["C7"] = "=SUM(C2:C6)"
aba_ativa["D7"] = "=SUM(D2:D6)"
print("=======DADOS DA PLANILHA EXCEL=======")

for linha in aba_ativa.iter_rows(values_only=True):
  print(linha)

wb.save("Automatizador_de_Comissões_de_Vendas.xlsx")
```
