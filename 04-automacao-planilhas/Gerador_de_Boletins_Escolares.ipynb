```python
from reportlab.pdfgen import canvas
alunos = [
    ["Matheus", 8, 6],
    ["Luam", 9, 7],
    ["Gabriela", 10, 5],
    ["Amanda", 6, 5],
    ["Juliana", 4, 5]
]
for linha in alunos:
  pdf = canvas.Canvas(f"Aluno_{linha[0]}.pdf")
  nota1 = linha[1]
  nota2 = linha[2]
  media = (nota1 + nota2) / 2

  pdf.drawString(100, 750, f"Boletinho do aluno: {linha[0]}")
  pdf.drawString(100, 720, f"Nota: {linha[1]} / {linha[2]}")
  pdf.drawString(100, 690, f"Média: {media}")

  print(f"As notas do {linha[0]} foram: {nota1} e {nota2}\nFicando com uma média de: {media}.")
  if media >= 7:
    pdf.drawString(100,660, f"Status: Aprovado! Parabéns")
    print(f"{linha[0]} está APROVADO! Parabéns")
    print('-'*35)
  elif media >= 5:
    pdf.drawString(100,660, f"Status: Recuperação! Atenção")
    print(f"{linha[0]} está de RECUPERAÇÃO! Cuidado!")
    print('-'*35)
  else:
    pdf.drawString(100,660, f"Status: Reprovado! Lamentavél")
    print(f"{linha[0]} está REPROVADO! Lamentamos")
    print('-'*35)

  pdf.save()
```
