```python
# Controle do clique do mouse
import pyautogui
import time
```

```python
# Contro de tempo entre um clique e outro clique
time.sleep(5)

# clique simples
pyautogui.click()

time.sleep(3)

# clique duplo
pyautogui.doubleClick()

time.sleep(3)

# clique com botão direito
pyautogui.rightClick()
```
