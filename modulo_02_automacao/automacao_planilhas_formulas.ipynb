```python
**Formatando cores de fundo e de texto**
```

```python
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill

wb = load_workbook("Relatorio Concluido.xlsx")
aba = wb.active

cor_fundo = PatternFill(start_color="000080", end_color="000080", fill_type="solid")

fonte_cabecalho = Font(name="Arial", size=11, bold=True, color="FFFFFF")

for col in range(1,5):
  celula = aba.cell(row=1, column=col)
  celula.fill = cor_fundo
  celula.font = fonte_cabecalho

wb.save("Relatorio Formatado.xlsx")
```

```python
**Fórmulas Matemáticas Automáticas**
```

```python
from openpyxl import load_workbook

# 1. Abre a planilha existente
wb = load_workbook("Relatorio Concluido.xlsx")
aba = wb.active

# 2. Descobre de forma dinâmica qual é a última linha preenchida
ultima_linha = aba.max_row
linha_do_total = ultima_linha + 2  # Deixa uma linha em branco para organizar

# 3. Escreve o texto indicador na Coluna 2 (Coluna B)
aba.cell(row=linha_do_total, column=2, value="TOTAL DE VENDAS:")

aba.cell(row=linha_do_total, column=3, value="=SUM(C2:C3)")


# 5. Salva as alterações no novo arquivo
wb.save("Relatorio_Com_Formulas.xlsx")

print("Robô de Fórmulas: Conta matemática injetada com sucesso!")
```
