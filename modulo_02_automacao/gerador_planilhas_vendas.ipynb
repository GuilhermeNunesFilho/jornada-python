```python
# O Gerador de Planilhas de Vendas

from openpyxl import Workbook

wb = Workbook()
aba = wb.active

aba.title = "Vendas_2026"

aba.append(["Nome do Cliente", "Modelo do Carro", "Valor", "Status"])

aba.append(["Cezar Algusto", "Onyx", "85000", "Aprovado"])
aba.append(["Guilherme Nunes", "Civic", "120000", "Aprovado"])
aba.append(["Sabrina Motta", "Corolla", "180000", "Aprovado"])

wb.save("Relatorio Concluido.xlsx")
print("Robô concluiu a missão")
```

```python
from openpyxl import load_workbook

wb = load_workbook("Relatorio Concluido.xlsx")
aba = wb.active

print(" === LENDO OS DADOS DA PLANILHA === ")
print("-"*35)

for linha in aba.iter_rows(values_only=True):
  print(linha)
```

```python
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill

wb = load_workbook("Relatorio Concluido.xlsx")
aba = wb.active

cor_fundo = PatternFill(start_color="000080", end_color="000080", fill_type="solid")

fonte_cabecalho = Font(name="Arial", size=11, bold=True, color="FFFFFF")

for col in range(1,5):
  celula = aba.cell(row=1, column=col)
  celula.fill = cor_fundo
  celula.font = fonte_cabecalho

wb.save("Relatorio Formatado.xlsx")
```
