```python
# importando as bibliotecas pywhatkit, time, pyautogui

import pywhatkit as kit
import time
import pyautogui

# criando a função que envia a mensagem
def enviar_mensagem_automatica():
    print("Iniciando o bot de mensagens...")
    
    # 1. Defina o número para enviar a menssagem
    
    numero_destino = "(xx)xxxxx-xxxx" 
    
    # 2. Defina a mensagem que o bot vai enviar
    
    mensagem = "Automatizando chatbot 🤖"
    
    # 3. O comando abre o WhatsApp Web e digita a mensagem automaticamente
    # Ele agenda para enviar daqui a 1 ou 2 minutos no horário do seu computador
    
    print("Abrindo o WhatsApp Web no navegador...")
    kit.sendwhatmsg_instantly(numero_destino, mensagem, wait_time=15)
    
    # 4. Simula o clique do teclado para enviar (Garante o envio)
    
    time.sleep(3)
    pyautogui.press('enter')
    print("Mensagem enviada com sucesso!")

# Executar a função
enviar_mensagem_automatica()
```
