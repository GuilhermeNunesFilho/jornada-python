```python
import pyautogui
import time
import datetime
```

```python
# criando variavel para marca o horário da inicialização
hora_inicio = datetime.time(15,3)

# criando variável para saber a hora atual
hora_agora = datetime.datetime.now()

# criando um contador para a inicialização
tempo_para_iniciar = datetime.datetime.combine(hora_agora.date(), hora_inicio)

# criando a crontro de tempo
segundos = (tempo_para_iniciar - hora_agora).total_seconds()

# criando a tarefa a ser executada
time.sleep(segundos)

pyautogui.write("Ola, estou aprendendo a automatizar", interval = 0.1)

time.sleep(2)
pyautogui.press("enter")
time.sleep(2)

pyautogui.write("xD")
```
