```python
import pyautogui
import time
```

```python
# Tempo para inciar a TAREFA
time.sleep(6)

# Buscando a posição do mouse atual
posicao = pyautogui.position()
print(f"A posição atual do mouse é: {posicao}")
```

```python
time.sleep(6)
# Movendo o mouse até as cordenada da tela x=36, y=763
pyautogui.moveTo(x=36, y=763, duration = 1)

time.sleep(1)

# Selecionar o arquivo
pyautogui.click()

time.sleep(1)

# Copiando o arquivo com o Método = HOTKEY
pyautogui.hotkey("ctrl","c")

time.sleep(1)

# Abrindo o navegador
pyautogui.moveTo(x=52, y=543, duration = 1)

time.sleep(1)

pyautogui.doubleClick()

time.sleep(2)
```
